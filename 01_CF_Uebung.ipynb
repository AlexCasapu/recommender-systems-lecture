```python
# DEMO: Collaborative Filtering (User Based)
import pandas as pd

# Spiel-Datensatz: Bewertungen von 1-5 Sternen
data = {
    'Item1': [5, 3, 4, 3, 1],
    'Item2': [3, 1, 5, 3, 5],
    'Item3': [4, 2, 4, 1, 5]
}
ratings = pd.DataFrame(data, index=['Alice', 'User1', 'User2', 'User3', 'User4'])
print("Rating Matrix:")
print(ratings)

# Pearson Korrelation berechnen
similarity_matrix = ratings.T.corr()
print("\nÄhnlichkeitsmatrix der Nutzer:")
print(similarity_matrix)

# Ähnlichste Nutzer zu Alice finden (ohne Alice selbst)
alice_similarities = similarity_matrix['Alice'].drop('Alice').sort_values(ascending=False)
print(f"\nÄhnlichster Nutzer zu Alice ist: {alice_similarities.index[0]} mit Score {alice_similarities.iloc[0]:.2f}")
```
